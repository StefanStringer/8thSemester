```python
import numpy as np
import matplotlib.pyplot as plt

k = 8.617e-5
T = np.linspace(200, 600, 500) 

materials = {
    'Silicon': {'Eg': 1.12, 'B': 5.29e15},
    'Germanium': {'Eg': 0.66, 'B': 1.66e15},
    'GaAs': {'Eg': 1.42, 'B': 2.10e14}
}

plt.figure(figsize=(10, 6))

for name, params in materials.items():
    Eg = params['Eg']
    B = params['B']
    
    ni = B * (T ** 1.5) * np.exp(-Eg / (2 * k * T))
    
    plt.semilogy(T, ni, label=name)

plt.title('Intrinsic Carrier Concentration v Temp', fontsize=14)
plt.xlabel('Temperature (K)', fontsize=12)
plt.ylabel('Intrinsic Carrier Concentration ', fontsize=12)
plt.grid(True, which="both", ls="-", alpha=0.6)
plt.legend(fontsize=12)
plt.show()
```
